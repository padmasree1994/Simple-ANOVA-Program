```python
#1.Simple ANOVA Example

#Compare marks of students from 3 different classes.

#Step 1: Install Libraries
!pip install scipy numpy matplotlib

#Step 2: Import Libraries
import numpy as np
from scipy.stats import f_oneway
import matplotlib.pyplot as plt

#Step 3: Create Student Marks Data
# Marks of 3 classes

class_A = [75, 80, 85, 90, 95]
class_B = [70, 72, 78, 74, 76]
class_C = [88, 85, 91, 89, 87]

print("Class A:", class_A)
print("Class B:", class_B)
print("Class C:", class_C)

#Step 4: Perform ANOVA Test
# ANOVA Test

f_statistic, p_value = f_oneway(
    class_A,
    class_B,
    class_C
)

print("F-Statistic:", f_statistic)
print("P-Value:", p_value)

#Step 5: Interpretation
if p_value < 0.05:
    print("Significant difference exists between groups")
else:
    print("No significant difference between groups")

#Step 6: Visualization
# Boxplot visualization

data = [class_A, class_B, class_C]

plt.boxplot(data)

plt.title("ANOVA Analysis")
plt.xlabel("Classes")
plt.ylabel("Marks")

plt.show()

#2. Predictive Analytics – Simple Explanation

#Step 1: Install Libraries
!pip install pandas matplotlib scikit-learn

#Step 2: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression

#Step 3: Create Dataset
# Study hours and marks dataset

data = {
    'Hours': [1,2,3,4,5,6,7,8],
    'Marks': [35,40,50,55,60,70,80,90]
}

# Create DataFrame

df = pd.DataFrame(data)

print(df)

#Step 4: Visualization
plt.scatter(
    df['Hours'],
    df['Marks'],
    color='blue'
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()

#Step 5: Prepare Data
# Input feature

X = df[['Hours']]

# Output target

y = df['Marks']

#Step 6: Train Predictive Model
# Create model

model = LinearRegression()

# Train model

model.fit(X, y)

print("Model Training Completed")

#Step 7: Predict Future Value
# Predict marks for 9 study hours

prediction = model.predict([[9]])

print("Predicted Marks:", prediction[0])

#Step 8: Predictive Line Visualization
# Prediction line

plt.scatter(
    df['Hours'],
    df['Marks'],
    color='red'
)

plt.plot(
    df['Hours'],
    model.predict(X)
)

plt.title("Predictive Analytics")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```
